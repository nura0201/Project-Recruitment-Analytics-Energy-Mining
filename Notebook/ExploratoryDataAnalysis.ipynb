```python
# Visualization 1: Distribusi Status Kandidat apakah kandidat merupakan mahasiswa atau fresh graduated
import matplotlib.pyplot as plt

status = df["Apakah Anda Memiliki Pengalaman Lebih Dari 1 Tahun / Fresh Graduate"].value_counts()

fig, ax = plt.subplots(figsize=(6,5))

bars = ax.bar(status.index, status.values)

ax.set_title("Apakah Kandidat Berpengalaman / Freshgraduated?")
ax.set_ylabel("Jumlah Kandidat")

# Menampilkan jumlah di atas setiap bar
ax.bar_label(bars, padding=3)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 2: Bar chart top 10 untuk kategori domisili
top10_domisili = df["Domisili"].value_counts().head(10)

fig, ax = plt.subplots(figsize=(8,5))

bars = top10_domisili.sort_values().plot(
    kind="barh",
    ax=ax
)

ax.set_title("Top 10 Domisili Kandidat")
ax.set_xlabel("Jumlah Kandidat")
ax.set_ylabel("")

# Menampilkan jumlah pada setiap bar
ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python
# Visualization 3: Bar chart top 5 untuk kategori skill
top5_skill = df["Kemampuan Karyawan"].value_counts().head(5)

fig, ax = plt.subplots(figsize=(8,5))

bars = top5_skill.sort_values().plot(
    kind="barh",
    ax=ax
)

ax.set_title("Top 5 Kemampuan Kandidat")
ax.set_xlabel("Jumlah Kandidat")
ax.set_ylabel("")

# Menampilkan jumlah pada setiap bar
ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python
# Visualization 4: Bar chart top 5 untuk kategori target masuk perusahaan
top5_target = df["Target Masuk Perusahaan"].value_counts().head(5)

fig, ax = plt.subplots(figsize=(8,5))

bars = top5_target.sort_values().plot(
    kind="barh",
    ax=ax
)

ax.set_title("Top 5 Target Masuk Perusahaan")
ax.set_xlabel("Jumlah Kandidat")
ax.set_ylabel("")

# Menampilkan jumlah pada setiap bar
ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python
# Visualization 5: Timeline Pendaftaran Kandidat
import matplotlib.pyplot as plt

# Pastikan Timestamp bertipe datetime
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Jumlah pendaftaran per tanggal
timeline = df.groupby(df["Timestamp"].dt.date).size()

plt.figure(figsize=(10,5))

plt.plot(
    timeline.index,
    timeline.values,
    marker="o",
    linewidth=2
)

plt.title("Timeline Pendaftaran Kandidat")
plt.xlabel("Tanggal")
plt.ylabel("Jumlah Kandidat")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
