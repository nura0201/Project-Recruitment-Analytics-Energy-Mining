```python
# Visualization 1: Distribusi Status Kandidat apakah kandidat merupakan mahasiswa atau fresh graduated
import matplotlib.pyplot as plt

status = df["Apakah Anda Memiliki Pengalaman Lebih Dari 1 Tahun / Fresh Graduate"].value_counts()

fig, ax = plt.subplots(figsize=(6,5))

bars = ax.bar(status.index, status.values)

ax.set_title("Apakah Kandidat Berpengalaman / Freshgraduated?")
ax.set_ylabel("Jumlah Kandidat")

# Menampilkan jumlah di atas setiap bar
ax.bar_label(bars, padding=3)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualization 2: Bar chart top 10 untuk kategori domisili
top10_domisili = df["Provinsi"].value_counts().head(10)

fig, ax = plt.subplots(figsize=(8,5))

bars = top10_domisili.sort_values().plot(
    kind="barh",
    ax=ax
)

ax.set_title("Candidate Distribution by Province")
ax.set_xlabel("")
ax.set_ylabel("")

# Menampilkan jumlah pada setiap bar
ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```

```python
# Visualization 3: Candidate Skills
import matplotlib.pyplot as plt

# Daftar kolom skill
skill_cols = [
    "Excel",
    "SQL",
    "Python",
    "Power BI",
    "Tableau", 
    "Looker Studio",
    "Databases"
]

# Hitung jumlah kandidat per skill
skill_counts = df[skill_cols].sum().sort_values(ascending=False)

# Membuat grafik
plt.figure(figsize=(8,5))
ax = skill_counts.plot(kind="bar")

# Menambahkan angka di atas bar
for container in ax.containers:
    ax.bar_label(container, fmt='%d', fontsize=10)

plt.title("Top Candidate Skills")
plt.xlabel("")
plt.ylabel("")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Visualization 4: Target masuk perusahaan
# Daftar kolom target
target_cols = [
    "Networking",
    "Upskill",
    "Cari Kerja",
    "Switch Career"
]

# Hitung jumlah kandidat
target_counts = df[target_cols].sum().sort_values(ascending=False)

# Plot
plt.figure(figsize=(8,5))
ax = target_counts.plot(kind="bar")

# Tambahkan angka di atas bar
for container in ax.containers:
    ax.bar_label(container, fmt='%d', fontsize=10)

plt.title("Target Company Goals")
plt.xlabel("")
plt.ylabel("Number of Candidates")
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

```python
# Visualization 5: Timeline Pendaftaran Kandidat

# Pastikan Timestamp bertipe datetime
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Jumlah pendaftaran per tanggal
timeline = df.groupby(df["Timestamp"].dt.date).size()

plt.figure(figsize=(10,5))

plt.plot(
    timeline.index,
    timeline.values,
    marker="o",
    linewidth=2
)

plt.title("Timeline Pendaftaran Kandidat")
plt.xlabel("Tanggal")
plt.ylabel("Jumlah Kandidat")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
