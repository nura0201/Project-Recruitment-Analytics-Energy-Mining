```python
# Distribusi Satus Kandidat apakah kandidat merupakan mahasiswa atau fresh graduated
import matplotlib.pyplot as plt

status = df["Apakah Anda Memiliki Pengalaman Lebih Dari 1 Tahun / Fresh Graduate"].value_counts()

fig, ax = plt.subplots(figsize=(6,5))

bars = ax.bar(status.index, status.values)

ax.set_title("Apakah Kandidat Berpengalaman / Freshgraduated?")
ax.set_xlabel("Status Kandidat")
ax.set_ylabel("Jumlah Kandidat")

# Menampilkan jumlah di atas setiap bar
ax.bar_label(bars, padding=3)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Bar chart top 5 untuk kategori domisili
top5_domisili = df["Domisili"].value_counts().head(5)

fig, ax = plt.subplots(figsize=(8,5))

bars = top5_domisili.sort_values().plot(
    kind="barh",
    ax=ax
)

ax.set_title("Top 5 Domisili Kandidat")
ax.set_xlabel("Jumlah Kandidat")
ax.set_ylabel("")

# Menampilkan jumlah pada setiap bar
ax.bar_label(ax.containers[0], padding=3)

plt.tight_layout()
plt.show()
```
