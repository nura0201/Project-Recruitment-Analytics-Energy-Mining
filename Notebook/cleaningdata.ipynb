4. Data Preparation

```python
# cek kolom nama lengkap
print(df["Nama Lengkap"].head(20))
```

```python
# cek nama lengkap yang formatnya kapital semua
df[df["Nama Lengkap"] == df["Nama Lengkap"].str.upper()]["Nama Lengkap"]
```

```python
# ubah format Nama Lengkap
df["Nama Lengkap"] = df["Nama Lengkap"].str.strip().str.title()
```

```python
# Mengganti nama kolom sesuai ketentuan
df = df.rename(columns={
    "Tools yang Ingin Dipelajari": "Kemampuan Karyawan",
    "Target Join Komunitas": "Target Masuk Perusahaan",
    "Tau Seara Data Dari Mana?": "Mengetahui Perusahaan Dari Mana?",
    "Apakah Anda Mahasiswa / Fresh Graduate": "Apakah Anda Memiliki Pengalaman Lebih Dari 1 Tahun / Fresh Graduate"
})
```

```python
# menghapus kolom yang tidak digunakan
df = df.drop(columns=["Pekerjaan & Instansi Saat Ini", "Column 1", "Domisili"])
```

```python
# ubah tipe data timestamp jadi date
df["Timestamp"] = pd.to_datetime(df["Timestamp"]).dt.normalize()
```

```python
df.info()
```

```python
print("\nKolom Setelah Preparation")
print(df.columns.tolist())
```

5. Data Cleaning

```python
# Menghapus spasi di awal dan akhir pada seluruh kolom kategorikal
string_cols = df.select_dtypes(include=["object", "string"]).columns

df[string_cols] = df[string_cols].apply(lambda col: col.str.strip())
```

```python
df = df.drop_duplicates(
    subset=["Nama Lengkap", "Email", "Nomor WhatsApp"]
)

print(len(df))
```

```python
# cek lagi duplikat by nama lengkap
duplicate_name = df[df.duplicated(subset="Nama Lengkap", keep=False)]
duplicate_name.sort_values("Nama Lengkap")
```

```python
# Menghapus data duplikat by Nama Lengkap
df = df.drop_duplicates(subset="Nama Lengkap", keep="first")

print(len(df))
```

```python
df.info()
```

```python
# menghapus kolom yang memuat informasi pribadi
df = df.drop(columns=["Nomor WhatsApp", "Email"])
```

```python
# Cek apakah ada missing value
df.isnull().sum()
```

```python
# simpan data yang sudah menyelesaikan tahap cleaning dengan format csv
df.to_csv("Database_Seara_Data_Clean.csv", index=False)
```
